```python
import random

# ============================================
# USER INPUT
# ============================================

GRID_SIZE = int(input("Enter grid size: "))
ITERATIONS = int(input("Enter number of iterations: "))

MIN_VEHICLES = float(input("Enter minimum number of vehicles: "))
MAX_VEHICLES = float(input("Enter maximum number of vehicles: "))


# ============================================
# FITNESS FUNCTION
# f(x) = 2x² - 20x + 70
# ============================================

def fitness(x):
    return 2 * x**2 - 20 * x + 70


# ============================================
# INITIALIZE POPULATION
# ============================================

grid = [
    [
        random.uniform(MIN_VEHICLES, MAX_VEHICLES)
        for _ in range(GRID_SIZE)
    ]
    for _ in range(GRID_SIZE)
]


# ============================================
# GET 3 × 3 NEIGHBORHOOD
# ============================================

def get_neighbors(grid, row, col):

    neighbors = []

    for i in range(row - 1, row + 2):

        for j in range(col - 1, col + 2):

            # Do not include current cell
            if i == row and j == col:
                continue

            # Check boundaries
            if 0 <= i < GRID_SIZE and 0 <= j < GRID_SIZE:
                neighbors.append(grid[i][j])

    return neighbors


# ============================================
# PARALLEL CELLULAR ALGORITHM
# ============================================

for iteration in range(ITERATIONS):

    # Create new grid for synchronous update
    new_grid = [
        row[:] for row in grid
    ]

    # Process every cell
    for i in range(GRID_SIZE):

        for j in range(GRID_SIZE):

            current_value = grid[i][j]

            # Get neighboring cells
            neighbors = get_neighbors(
                grid,
                i,
                j
            )

            if neighbors:

                # Find neighbor with minimum fitness
                best_neighbor = min(
                    neighbors,
                    key=fitness
                )

                # If neighbor is better
                if fitness(best_neighbor) < fitness(current_value):

                    # Move halfway toward better solution
                    new_value = (
                        current_value + best_neighbor
                    ) / 2

                    # Keep within allowed range
                    new_value = max(
                        MIN_VEHICLES,
                        min(MAX_VEHICLES, new_value)
                    )

                    new_grid[i][j] = new_value

                else:
                    new_grid[i][j] = current_value

    # Update entire grid
    grid = new_grid


# ============================================
# FIND BEST SOLUTION
# ============================================

best_x = grid[0][0]

for row in grid:

    for x in row:

        if fitness(x) < fitness(best_x):
            best_x = x


best_cost = fitness(best_x)


# ============================================
# OUTPUT
# ============================================

print("\n--------------------------------------")
print("PARALLEL CELLULAR ALGORITHM")
print("--------------------------------------")

print("Grid Size:", GRID_SIZE, "x", GRID_SIZE)
print("Number of Cells:", GRID_SIZE * GRID_SIZE)
print("Iterations:", ITERATIONS)

print()
print("Best Number of Vehicles:",
      round(best_x, 2))

print("Minimum Estimated Cost:",
      round(best_cost, 2))

print("--------------------------------------")

```
