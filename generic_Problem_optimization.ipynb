```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the problem
# Maximize f(x) = x * sin(10*pi*x) + 1
def fitness(x):
    return x * np.sin(10 * np.pi * x) + 1


# 2. Initialize GA parameters
population_size = 20
mutation_rate = 0.15
crossover_rate = 0.8
generations = 50

# Search range
lower_bound = 0
upper_bound = 1


# 3. Create initial population
population = np.random.uniform(
    lower_bound,
    upper_bound,
    population_size
)


best_fitness_history = []
average_fitness_history = []

# 4-8. Run the Genetic Algorithm
for generation in range(generations):

    # 4. Evaluate fitness
    fitness_scores = fitness(population)

    best_index = np.argmax(fitness_scores)
    best_fitness = fitness_scores[best_index]
    average_fitness = np.mean(fitness_scores)

    best_fitness_history.append(best_fitness)
    average_fitness_history.append(average_fitness)

    # 5. Selection
    # Select the top 50% individuals
    sorted_indices = np.argsort(fitness_scores)[::-1]
    selected = population[sorted_indices[:population_size // 2]]

    # 6. Crossover
    new_population = list(selected)

    while len(new_population) < population_size:

        parent1, parent2 = np.random.choice(selected, 2, replace=False)

        if np.random.rand() < crossover_rate:
            alpha = np.random.rand()
            child = alpha * parent1 + (1 - alpha) * parent2
        else:
            child = parent1

        # 7. Mutation
        if np.random.rand() < mutation_rate:
            child += np.random.normal(0, 0.05)

        # Keep child within bounds
        child = np.clip(child, lower_bound, upper_bound)

        new_population.append(child)

    population = np.array(new_population)

    print(
        f"Generation {generation + 1}: "
        f"Best Fitness = {best_fitness:.6f}, "
        f"Best x = {population[best_index]:.6f}"
    )


# 9. Output the best solution
final_fitness = fitness(population)
best_index = np.argmax(final_fitness)

best_x = population[best_index]
best_value = final_fitness[best_index]

print("\nBest Solution Found:")
print(f"x = {best_x:.6f}")
print(f"Maximum value = {best_value:.6f}")

```
