```python

import numpy as np

# -------------------------------------------------
# Scenario: Optimize two operating parameters
# Goal: Find settings that minimize operating loss
# -------------------------------------------------

def objective_function(X):
    """
    X[0] = operating parameter 1
    X[1] = operating parameter 2

    Illustrative loss function with an optimum at
    parameter 1 = 3 and parameter 2 = -2.
    """
    return (X[0] - 3)**2 + (X[1] + 2)**2


def grey_wolf_optimizer(
    objective_function,
    dim=2,
    n_wolves=20,
    max_iter=100,
    lb=-10,
    ub=10
):
    # 1. Initialize grey wolf population Xi
    X = np.random.uniform(lb, ub, (n_wolves, dim))

    # 2. Initialize alpha, beta, and delta
    X_alpha = np.zeros(dim)
    X_beta = np.zeros(dim)
    X_delta = np.zeros(dim)

    f_alpha = float("inf")
    f_beta = float("inf")
    f_delta = float("inf")

    # 3. Calculate initial fitness and rank wolves
    for i in range(n_wolves):
        fitness = objective_function(X[i])

        if fitness < f_alpha:
            f_delta, X_delta = f_beta, X_beta.copy()
            f_beta, X_beta = f_alpha, X_alpha.copy()
            f_alpha, X_alpha = fitness, X[i].copy()

        elif fitness < f_beta:
            f_delta, X_delta = f_beta, X_beta.copy()
            f_beta, X_beta = fitness, X[i].copy()

        elif fitness < f_delta:
            f_delta, X_delta = fitness, X[i].copy()

    # 4. Main optimization loop
    for t in range(max_iter):

        # Decrease a linearly from 2 to 0
        a = 2 - 2 * t / max_iter

        # Update each search agent
        for i in range(n_wolves):
            for j in range(dim):

                # Alpha wolf influence
                r1, r2 = np.random.rand(), np.random.rand()
                A1 = 2 * a * r1 - a
                C1 = 2 * r2
                D_alpha = abs(C1 * X_alpha[j] - X[i, j])
                X1 = X_alpha[j] - A1 * D_alpha

                # Beta wolf influence
                r1, r2 = np.random.rand(), np.random.rand()
                A2 = 2 * a * r1 - a
                C2 = 2 * r2
                D_beta = abs(C2 * X_beta[j] - X[i, j])
                X2 = X_beta[j] - A2 * D_beta

                # Delta wolf influence
                r1, r2 = np.random.rand(), np.random.rand()
                A3 = 2 * a * r1 - a
                C3 = 2 * r2
                D_delta = abs(C3 * X_delta[j] - X[i, j])
                X3 = X_delta[j] - A3 * D_delta

                # Update wolf position
                X[i, j] = (X1 + X2 + X3) / 3

            # Keep wolves inside search boundaries
            X[i] = np.clip(X[i], lb, ub)

        # 5. Recalculate fitness and update leaders
        for i in range(n_wolves):
            fitness = objective_function(X[i])

            if fitness < f_alpha:
                X_delta, f_delta = X_beta.copy(), f_beta
                X_beta, f_beta = X_alpha.copy(), f_alpha
                X_alpha, f_alpha = X[i].copy(), fitness

            elif fitness < f_beta:
                X_delta, f_delta = X_beta.copy(), f_beta
                X_beta, f_beta = X[i].copy(), fitness

            elif fitness < f_delta:
                X_delta, f_delta = X[i].copy(), fitness

        # Display progress
        if (t + 1) % 10 == 0:
            print(
                f"Iteration {t + 1}: "
                f"Best fitness = {f_alpha:.8f}"
            )

    # 6. Return the best search agent
    return X_alpha, f_alpha


# -------------------------------------------------
# Run the algorithm
# -------------------------------------------------

np.random.seed(42)  # Reproducible results

best_solution, best_fitness = grey_wolf_optimizer(
    objective_function=objective_function,
    dim=2,
    n_wolves=20,
    max_iter=100,
    lb=-10,
    ub=10
)

print("\nOptimization completed!")
print("Best operating parameters:")
print(f"Parameter 1 = {best_solution[0]:.6f}")
print(f"Parameter 2 = {best_solution[1]:.6f}")
print(f"Minimum operating loss = {best_fitness:.10f}")
```
